### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()

# Lista de URLs y nombres
urls = [url, url2, url3, url4]
nombres = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

# Crear df_total unificado con columna "Tienda"
dataframes = []
for i in range(4):
    df = pd.read_csv(urls[i])
    df["Tienda"] = nombres[i]
    dataframes.append(df)

df_total = pd.concat(dataframes, ignore_index=True)

# Calcular facturación base
df_total["Facturación"] = df_total["Precio"] + df_total["Costo de envío"]
```

#1. Análisis de facturación



```python
# Facturación total y por tienda
print("✅ Facturación total:", df_total["Facturación"].sum())
print("\n📊 Facturación por tienda:")
print(df_total.groupby("Tienda")["Facturación"].sum())
```

# 2. Ventas por categoría

```python
ventas_categoria = df_total.groupby("Categoría del Producto").agg({
    "Facturación": "sum",
    "Producto": "count"
}).rename(columns={
    "Facturación": "Total Facturado",
    "Producto": "Cantidad de Ventas"
}).sort_values(by="Total Facturado", ascending=False)

print("📊 Ventas por categoría:")
print(ventas_categoria)
```

# 3. Calificación promedio de la tienda


```python
calificacion_promedio = df_total.groupby("Tienda")["Calificación"].mean().round(2)
print("⭐ Calificación promedio por tienda:")
print(calificacion_promedio)

promedio_general = df_total["Calificación"].mean().round(2)
print(f"\n⭐ Calificación promedio total: {promedio_general}")
```

# 4. Productos más y menos vendidos

```python
ventas_por_producto = df_total["Producto"].value_counts()

print("🔥 Productos más vendidos:")
print(ventas_por_producto.head(10))

print("\n🧊 Productos menos vendidos:")
print(ventas_por_producto.tail(10))
```

# 5. Envío promedio por tienda

```python
envio_promedio = df_total.groupby("Tienda")["Costo de envío"].mean().round(2)

print("🚚 Envío promedio por tienda:")
print(envio_promedio)
```
